# 🚢 Titanic Survival Prediction

## Project Goal
This project uses the well-known Titanic dataset (from Kaggle) to build a machine learning model that predicts whether a passenger survived, based on features such as age, sex, ticket class, number of family members aboard, and more.

## Steps Performed
1. Loading and exploring the data (EDA)
2. Cleaning the data and handling missing values
3. Feature engineering: extracting `Title`, `Deck`, `FamilySize`, and `IsAlone`
4. Encoding categorical features into numeric form
5. Splitting the data into training and test sets
6. Training and evaluating classification models (Logistic Regression, Decision Tree)

```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\ARIO\Documents\22 ML project\Titanic Survival Prediction\Titanic-Dataset.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df['Parch'].value_counts()
```

```python
df['Embarked'].mode()[0]
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['Title'] = df['Name'].str.extract(' ([A-Za-z]+)\.')
```

```python
df['Title'].value_counts()
```

```python
df['Title'] = df['Title'].replace({'Mlle':'Miss' , 'Mme':'Mr' ,'Ms':'Miss'})
```

```python
df['Title'].value_counts()
```

```python
rare_titles = ['Dr','Rev','Major','Col','Don','Lady','Sir','Capt','Countess','Jonkheer']
df['Title'] = df['Title'].replace(rare_titles, 'Rare')
```

```python
df['Title'].value_counts()
```

```python
df['Age'] = df['Age'].fillna(df.groupby('Title')['Age'].transform('median'))
```

```python
df['Age'].head()
```

```python
df.isnull().sum()
```

```python
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

```python
df['Cabin'].value_counts()
```

```python
df['Deck'] = df['Cabin'].str[0]
```

```python
df = df.drop('Cabin', axis=1)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
```

```python
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
```

```python
df.head()
```

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
```

```python
df['Sex'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)
```

```python
df = pd.get_dummies(df, columns=['Title', 'Deck'], drop_first=True)
```

```python
df.head()
```

```python
cdf = df.drop(['Name','PassengerId','Ticket'], axis=1)
```

```python
cdf.head()
```

```python
X = cdf.drop(['Survived'] , axis=1)
y = cdf['Survived']
```

```python
from sklearn.model_selection import train_test_split 
X_train, X_test, y_train, y_test  = train_test_split(X,y,test_size=0.2,random_state=4)
```

```python
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train,y_train)
```

```python
y_hat = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_hat)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, y_hat)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_hat))
```

```python
from sklearn.tree import DecisionTreeClassifier 
```

```python
dt_model2 = DecisionTreeClassifier(max_depth=4, random_state=4)
dt_model2.fit(X_train, y_train)

print("Train:", dt_model2.score(X_train, y_train))
print("Test:", dt_model2.score(X_test, y_test))
```

```python
new_passenger = pd.DataFrame(0, index=[0], columns=X_train.columns)
```

```python
new_passenger['Pclass'] = 1
new_passenger['Sex'] = 1
new_passenger['Age'] = 30
new_passenger['Fare'] = 80
new_passenger['FamilySize'] = 1
new_passenger['IsAlone'] = 1
new_passenger['Embarked_S'] = 1
new_passenger['Title_Mrs'] = 1
```

```python
model.predict(new_passenger)
```

## Summary & Results

Two models were trained and compared on the test set:

| Model | Train Accuracy | Test Accuracy |
|---|---|---|
| Logistic Regression | – | 86.0% |
| Decision Tree (no depth limit) | 98.6% | 81.0% |
| Decision Tree (max_depth=4) | 83.4% | 84.9% |

### Key Takeaways
- Feature engineering (especially extracting `Title` from `Name`, and building `FamilySize` / `IsAlone`) had a meaningful positive impact on model performance.
- The unrestricted Decision Tree overfit the training data (98.6% train vs. 81.0% test accuracy), showing a large gap between train and test performance.
- Limiting `max_depth` to 4 significantly reduced overfitting and improved generalization to unseen data.
- Logistic Regression achieved the best overall test accuracy (86.0%) among the models tried.

### Possible Next Steps
- Try a Random Forest classifier, which typically outperforms a single Decision Tree.
- Apply feature scaling to improve Logistic Regression's convergence and performance.
- Use cross-validation for a more robust estimate of model performance.
- Tune hyperparameters (e.g. `max_depth`, `min_samples_leaf`) more systematically with GridSearchCV.
